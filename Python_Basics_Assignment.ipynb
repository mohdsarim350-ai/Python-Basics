## Question 2: Difference between List and Tuple in Python

| List | Tuple |
|---|---|
| Mutable – elements can be changed | Immutable – elements cannot be changed |
| Written using `[]` | Written using `()` |
| Generally used when data may change | Useful for fixed/constant collections |
| Provides methods such as `append()` and `remove()` | Has fewer modification methods |

### Example

```python
my_list = [10, 20, 30]
my_tuple = (10, 20, 30)

my_list[0] = 100
print("Modified list:", my_list)
print("Tuple:", my_tuple)
```

## Question 3: What is a function in Python? Why are functions useful?

A **function** is a reusable block of code designed to perform a particular task. It is defined using the `def` keyword.

### Functions are useful because:
- They reduce code repetition.
- They make programs easier to organize.
- They improve readability.
- They make code easier to test and maintain.
- The same logic can be reused with different inputs.

## Question 5: Check whether a number is even or odd using conditional statements

```python
number = int(input("Enter a number: "))

if number % 2 == 0:
    print("The number is Even.")
else:
    print("The number is Odd.")
```

## Question 7: Create a list of five numbers and print the maximum number

```python
numbers = [12, 45, 7, 89, 34]

maximum_number = max(numbers)

print("List:", numbers)
print("Maximum number:", maximum_number)
```

## Question 8: Remove duplicate values from a list using a set

```python
numbers = [1, 2, 2, 3, 4, 4, 5, 5, 5]

unique_numbers = list(set(numbers))

print("Original list:", numbers)
print("List after removing duplicates:", unique_numbers)
```

## Question 9: Function that returns the square of a number

```python
def square(number):
    return number ** 2

print("Square of 6:", square(6))
```

## Question 10: Count how many times a number appears in a list

**Example list:** `[2, 3, 4, 2, 5, 2]`

```python
numbers = [2, 3, 4, 2, 5, 2]

number_to_count = int(input("Enter the number to count: "))
count = numbers.count(number_to_count)

print("List:", numbers)
print(f"{number_to_count} appears {count} time(s).")
```
